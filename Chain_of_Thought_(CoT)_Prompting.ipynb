Task 4 - Chain of Thought Prompting

A patient has
 * Fever
 * Cough
 * Oxygen Level = 88%

Design a prompt that encourages the LLM to reason step-by-step before providing the final recommendation

(Prompt design only; do not reveal the model's internal reasoning)

```python
!pip install transformers accelerate torch
```

```python
from transformers import pipeline
```

```python
generator = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct")
```

```python
prompt = """
You are an experienced medical assistant.

Context:
A patient presents with:
- Fever
- Cough
- Oxygen Level: 88%

Instructions:
1. Carefully analyze the patient's symptoms.
2. Assess the severity.
3. Provide a concise recommendation.
4. Recommend immediate medical attention if necessary.

Constraints:
- Do not reveal your internal reasoning.
- Do not diagnose a disease.
- Keep the response professional.

Output Format:

### Assessment

### Recommendation

### Disclaimer
"""
```

```python
response = generator(prompt, max_length=1000)
```

```python
print(response[0]["generated_text"])
```

In this task, we designed a Chain of Thought Prompt that guides the model to internally analyze patient symptoms before generating a final recommendation. By combining Role Prompting, Context, Constraints, and Structured Output, the prompt produces professional, organized, and safer responses while keeping the model's internal reasoning hidden, as required by the assignment.
