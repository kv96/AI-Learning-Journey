Task 3 - Few - Shot Prompting

Create a Few - Shot prompt that converts clinical notes into structured summaries.

Example:-

Input

* patient has cough and fever

Output

Symptoms:

* Cough

* Fever

Severity:

Moderate
Provides at least three examples before the final input.

```python
!pip install transformers accelerate torch
```

```python
from transformers import pipeline
```

```python
generator = pipeline("text-generation", model="Qwen/Qwen3-4B")
```

```python
prompt = """
You are an AI medical assistant.

Convert the following clinical notes into structured summaries.

Follow exactly the same format as the examples.

Example 1

Input:
Patient has cough and fever.

Output:

Symptoms:
- Cough
- Fever

Severity:
Moderate

-----------------------------------

Example 2

Input:
Patient complains of chest pain and shortness of breath.

Output:

Symptoms:
- Chest Pain
- Shortness of Breath

Severity:
High

-----------------------------------

Example 3

Input:
Patient reports headache and mild dizziness.

Output:

Symptoms:
- Headache
- Mild Dizziness

Severity:
Low

-----------------------------------

Now convert the following.

Input:
Patient has sore throat, fever and body pain.

Output:
"""
```

```python
response = generator(prompt, max_length=500)
```

```python
print(response[0]["generated_text"])
```

In this task, we used Few-Shot Prompting to teach an LLM how to convert clinical notes into structured medical summaries. By providing three examples before the final input, the model learned the desired response format and generated a consistent, organized output. This technique is especially useful when you want reliable results without fine-tuning the model.
