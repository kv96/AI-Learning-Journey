Installs the Hugging Face Transformers library and its dependencies in the Google Colab environment.

```python
from transformers import pipeline
import json
```

* pipeline is used to load the Hugging Face text-generation model.

* json is used to validate and format the generated JSON response.

```python
generator = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct" )
```

Loads an instruction-tuned language model capable of following structured prompts.

```python
prompt = """
You are an AI healthcare assistant.

Analyze the patient's symptoms.

Patient Information:
- Fever
- Dry Cough
- Fatigue

Return ONLY valid JSON.

Use this format:

{
  "disease": "",
  "symptoms": [],
  "risk_level": "",
  "recommendation": ""
}
"""
```

This prompt clearly specifies:

* The AI's role.

* The patient's symptoms.

* The required JSON format.

* The restriction to output only JSON.

```python
response = generator( prompt, max_new_tokens=350)
```

* max_new_tokens=350 limits the response length.

```python
print(response[0]["generated_text"])
```

Displays the generated JSON response.
