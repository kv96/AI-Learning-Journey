```python
from transformers import AutoTokenizer
from transformers import AutoModelForCausalLM
from transformers import pipeline
```

```python
model_name="Qwen/Qwen2.5-0.5B-Instruct"
```

Stores the model name that will be downloaded from **Hugging Face.**

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

Converts human-readable text into tokens that the model understands.

```python
model = AutoModelForCausalLM.from_pretrained(model_name)
```

Downloads the pretrained language model used for text generation.

```python
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)
```

Creates a text generation pipeline by combining the tokenizer and the model.

```python
prompt = "Patient has fever for three days."

response = generator(
    prompt,
    max_new_tokens=100
)

print(response[0]["generated_text"])
```

Accepts the user's question.
Generates the AI response.
Prints the generated output.

Successfully created a **healthcare chatbot using Hugging Face Transformers**. The chatbot can generate responses to medical questions using an instruction-tuned language model.
