Task 2:- Prompt Engineering

Objective :-
The objective of this task is to improve a poorly written prompt by applying Prompt Engineering techniques. A well-designed prompt helps the Large Language Model (LLM) generate accurate, structured, and reliable responses while reducing ambiguity and hallucinations.


Learning Outcomes:-
After completing this task, you will understand:

* What Prompt Engineering is.

* Why prompt quality affects AI responses.

* Role Prompting.

* Context-based Prompting.

* Constraints in prompts.

* Output Formatting using Markdown.



**Problem Statement**

Poor Prompt :-

* Tell me about diabetes.

Why is this prompt poor?

This prompt is too short and vague because:

* It does not specify who the AI should act as.

* No target audience is mentioned.

* No response format is specified.

* Important topics are missing.

* The model may generate inconsistent or unstructured answers.

**Prompt Engineering Techniques Used**


1. **Role Prompting**

Assign a role to the AI.

Example:

You are an experienced medical doctor.

Why?

Giving the AI a role helps it generate responses from the perspective of that profession, leading to more focused and relevant answers.

2. **Context**

Provide background information.

Example:

The user wants basic medical information for educational purposes.

Why?

Context helps the model understand the purpose of the response and avoid unnecessary or irrelevant information.

3. **Constraints**

Specify what the model should and should not do.

Example:

* Use simple language.

* Do not provide a medical diagnosis.

* Do not recommend medications.

* Suggest consulting a doctor when necessary.

Why?

Constraints make the response safer, more reliable, and appropriate for the intended audience.

4. Output Formatting

Tell the model exactly how to organize its response.

Example:

Return the answer in Markdown with headings and bullet points.

Why?

Structured output is easier to read and more professional.

```python
from transformers import pipeline
```

Imports the Hugging Face pipeline function, which provides an easy interface for text generation.

```python
generator = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct")
```

Loads the instruction-tuned Qwen model for generating responses based on prompts.

```python
#improved prompt

prompt = """
You are an experienced medical doctor and healthcare educator.

Context:
The user wants educational information about diabetes.

Instructions:
Generate the response in Markdown format.

Include:
1. Disease Overview
2. Symptoms
3. Risk Factors
4. Prevention
5. When to Consult a Doctor

Constraints:
- Use simple language.
- Do not provide medical diagnosis.
- Keep the answer concise.
- Add an educational disclaimer.
"""
```

Creates a detailed prompt using Role Prompting, Context, Constraints, and Output Formatting.

```python
response = generator(prompt, max_new_tokens=500)
```

```python
print(response[0]['generated_text'])
```
